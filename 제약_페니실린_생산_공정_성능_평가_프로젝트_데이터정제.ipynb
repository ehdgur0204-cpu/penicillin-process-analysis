```python
df_33col = df.iloc[:, :33]
```

```python
# 배치 시작 지점(즉, 시간이 0.2인 시점)의 인덱스를 찾기
batch_start_indices = df_33col['Time (h)'].index[df_33col['Time (h)'] == 0.2]
batch_start_indices
```

```python
def find_batch_end_indices(df):
    """
    데이터프레임에서 `Time (h)` 컬럼이 0.2 시간 간격(새 배치의 시작)인 행을 기준으로
    이전 배치의 종료 인덱스를 찾기

    매개변수 (Args):
        df: 판다스(Pandas) 데이터프레임.

    반환값 (Returns):
        판다스 Series: 각 배치가 종료되는 시점의 행 인덱스를 담고 있음
        (원리: 다음 배치가 시작되는 0.2 시간의 바로 앞 인덱스를 계산함)
    """
    # 'Time (h)'가 0.2인 곳의 인덱스를 찾기. (새 배치의 시작점)
    indices_0_2 = df_33col[df_33col['Time (h)'] == 0.2].index

    # 0.2인 지점의 바로 앞 인덱스를 계산 (이전 배치의 종료점)
    indices_before_0_2 = indices_0_2 - 1

    # 음수 인덱스(데이터프레임 시작점보다 앞선 경우)를 필터링하여 제외
    indices_before_0_2_filtered = indices_before_0_2[indices_before_0_2 > 0]

    # 0보다 큰 필터링된 인덱스에 대한 마스크를 생성
    indices_before_0_2_filtered_mask = indices_before_0_2_filtered > 0

    # 배치의 종료 인덱스를 나타내는 필터링된 인덱스를 반환
    return indices_before_0_2_filtered[indices_before_0_2_filtered_mask]

# 함수를 호출(원본 손상을 막기 위해 df.copy() 사용)
batch_end_indices = find_batch_end_indices(df_33col.copy())

# 마지막 배치의 종료 시점을 나타내기 위해, 데이터프레임의 맨 마지막 인덱스를 추가
batch_end_indices = batch_end_indices.append(pd.Index([df_33col.index[-1]]))

# 구해진 배치 종료 인덱스(batch_end_indices)를 출력
print(batch_end_indices)
```

## 2.Statistics 데이터

```python
df_stat_path = '/content/drive/MyDrive/01.Project/스파르타_튜터링/100_Batches_IndPenSim_Statistics.csv'
df_stat = pd.read_csv(df_stat_path)
df_stat.head()
```

# 데이터 매핑

```python
import pandas as pd

df_process = df_33col.copy()
stats_df = df_stat.copy()

#  Process 데이터(V3)에 'Batch_ID' 부여하기
df_process['Batch_ID'] = 0  # 초기화

for batch_num, (start_idx, end_idx) in enumerate(zip(batch_start_indices, batch_end_indices), 1):
    # 해당 구간(start~end)의 모든 행에 배치 번호를 입력
    df_process.loc[start_idx:end_idx, 'Batch_ID'] = batch_num

# Statistics 데이터는 순서대로 1~100번 배치의 정보이므로 ID를 생성
stats_df['Batch_ID'] = range(1, 101)

# df_process(시계열)를 기준으로 stats_df(결과값)를 붙이기
merged_df = pd.merge(df_process, stats_df, on='Batch_ID', how='left')


print(f"기존 데이터 크기: {df_process.shape}")
print(f"병합된 데이터 크기: {merged_df.shape}")
```
